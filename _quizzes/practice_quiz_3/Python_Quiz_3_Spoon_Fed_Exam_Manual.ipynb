## A — Running minimum: oldest person / smallest value
**Recognize:** “find oldest/youngest,” “find lowest/highest,” “running minimum/maximum,” or “find the name corresponding to a value.”

**Practice question:** Each person's name is an outer dictionary key; their birth year is an inner dictionary entry. The oldest person has the **smallest birth year**.

**Exact recipe:**
1. Set a very large initial year and a placeholder name.
2. Loop over each person's name.
3. Get that person's year.
4. If the year is smaller than the current minimum, update **both** year and name.
5. Print **after** the loop.

**Read:** `study_group[name]["year"]` = look up the dictionary for this person, then look up its `"year"` field.

**Change for maximum:** Use `>` rather than `<`, and start with `float("-inf")` (or another safely low value).

````python
study_group = {"Abed": {"year": 2007}, "Britta": {"year": 1998}, "Pierce": {"year": 1962}}
oldest_year = 3000
oldest_name = "nobody"

for name in study_group.keys():
    year = study_group[name]["year"]
    if year < oldest_year:
        oldest_year = year
        oldest_name = name

print(f"The oldest member is {oldest_name}, born in {oldest_year}")
# Pierce, 1962
````

## B — Nested loops: average for each day
**Recognize:** “average across cities/people for each day/category,” “using for loops,” “random subset.”

**Practice question:** `cities_to_be_averaged` contains a random selection of cities. Each city contains temperatures for days of the week. Calculate **one average per day**.

**Recipe:**
1. Outer loop chooses a day.
2. Set `temp_sum = 0` for that day.
3. Inner loop visits each *selected* city and adds that day's temperature.
4. After inner loop finishes, divide sum by number of selected cities.
5. Print the result, still inside the outer loop.

**Critical indentation:** `temp_sum = 0` is inside outer loop, before inner loop. Average calculation is outside inner loop.

**Do not divide by 5** if only 2–4 cities were selected. Use `len(cities_to_be_averaged)`.

**Variants:** For sum only, omit division. For 1 decimal place, use `round(x, 1)`. For average per city, swap which dimension the outer loop traverses.

````python
cities_to_be_averaged = {
    "Atlanta": {"Monday": 78, "Tuesday": 81},
    "Savannah": {"Monday": 84, "Tuesday": 86},
    "Athens": {"Monday": 76, "Tuesday": 79}
}
days_of_week = ["Monday", "Tuesday"]

for day in days_of_week:
    temp_sum = 0
    for city in cities_to_be_averaged.keys():
        temp_sum = temp_sum + cities_to_be_averaged[city][day]
    temp = round(temp_sum / len(cities_to_be_averaged), 2)
    print(f"Average high on {day}: {temp} °F")
# Monday 79.33; Tuesday 82.0
````

## C — Convert repeated calculations into a function with while
**Recognize:** “turn this code into a function,” “how many years until,” “grow until a target,” “return number of steps.”

**Practice question:** Starting population = 1000, multiply by `(1 + growth_rate)` annually, count years until population is at least 2000.

**Recipe:**
1. `def years_to_double(growth_rate):` declares one input.
2. Initialize population and counter **inside** the function so every call starts fresh.
3. `while population < 2000:` means repeat until population reaches 2000.
4. Update population **and** counter each time.
5. `return years` after the loop, but still indented within the function.
6. Call the function separately for each rate.

**Watch out:** `return` inside the `while` loop would stop after one year. An unchanged population causes an infinite loop. 1% is `0.01`, not `1`.

````python
def years_to_double(growth_rate):
    population = 1000
    years = 0
    while population < 2000:
        population = population * (1 + growth_rate)
        years = years + 1
    return years

for rate in [0.01, 0.02, 0.05, 0.10]:
    print(f"Growth rate {rate}: {years_to_double(rate)} years")
# Expected years: 70, 36, 15, 8
````

## E — try/except: skip invalid orders without crashing
**Recognize:** “skip invalid entries,” “dictionary key missing,” “invalid quantity,” “use try/except rather than if,” “print an error message but keep going.”

**Practice question:** `order` maps items to quantities stored as strings, e.g. `{"apple":"4","kiwi":"2","banana":"lots","cherry":"1"}`. `prices` maps valid items to prices.

**Recipe:**
1. Define `order_total(order, prices)`.
2. Set `total = 0` once.
3. Loop over each item in the order.
4. In `try`, get `prices[item]`, convert `order[item]` to integer, multiply, add.
5. If item is missing from `prices`, Python raises **KeyError**. Catch it and print “don't sell”.
6. If `int("lots")` fails, Python raises **ValueError**. Catch it and print “invalid quantity”.
7. Continue automatically to the next item. Return total after loop.

**Difference:** `KeyError` = missing dictionary key; `ValueError` = conversion fails; `TypeError` = wrong kind of object/operation.

**Note:** `int("3.5")` raises ValueError. `int("-2")` succeeds; the practice quiz does not separately forbid negative quantities.

**Common error:** putting `return total` inside the loop stops after the first item.

````python
prices = {"apple": 0.50, "banana": 1.25, "cherry": 3.00}
order = {"apple": "4", "kiwi": "2", "banana": "lots", "cherry": "1"}

def order_total(order, prices):
    total = 0
    for item in order.keys():
        try:
            total = total + prices[item] * int(order[item])
        except KeyError:
            print(f"Sorry, we don't sell {item}")
        except ValueError:
            print(f"Sorry, {order[item]} isn't a valid quantity")
    return total

total = order_total(order, prices)
print("Total:", total)
# Expected total: 5.0
````

## F — Tuples, polar coordinates, keyword-only argument
**Recognize:** “two points,” “tuple,” “Cartesian/polar,” “convert coordinates,” “keyword-only argument,” “default `polar=False`.”

**Practice question:** The given function `euclidean_dist(point1, point2)` computes distance for Cartesian `(x,y)` points. Write `cord_dist` to also accept polar `(r,theta)` points.

**Know this:**
- Tuple `(5,3)` contains two coordinates.
- `point[0]` = first number; `point[1]` = second.
- Polar conversion: `x = r * np.cos(theta)`; `y = r * np.sin(theta)`.
- NumPy trigonometric functions use **radians**.
- `def cord_dist(coord1, coord2, *, polar=False):` means first two arguments may be positional; `polar` must be written with its name when passed (`polar=True`). The `*` enforces this.
- When `polar=False`, leave points unchanged.
- When `polar=True`, convert **both** points before calculating distance.
- Return the result of the existing `euclidean_dist` function.

**Common mistake:** `cord_dist(p1,p2,True)` is invalid because `polar` is keyword-only. Use `cord_dist(p1,p2,polar=True)`.

````python
import numpy as np

def euclidean_dist(point1, point2):
    x1, y1 = point1[0], point1[1]
    x2, y2 = point2[0], point2[1]
    return np.sqrt((x1-x2)**2 + (y1-y2)**2)

def cord_dist(coord1, coord2, *, polar=False):
    if polar == True:
        coord1 = (coord1[0]*np.cos(coord1[1]), coord1[0]*np.sin(coord1[1]))
        coord2 = (coord2[0]*np.cos(coord2[1]), coord2[0]*np.sin(coord2[1]))
    return euclidean_dist(coord1, coord2)

print(cord_dist((1,1),(4,5)))        # 5.0
print(cord_dist((1.41,0.79),(6.4,0.9),polar=True))
````

## Extra Lecture 13 — assert, data validation, debugging
**The uploaded Lecture 13 notebook emphasizes checking data before trusting results.**
- `assert condition, "message"` raises `AssertionError` when condition is false.
- Use checks for valid ranges, valid types, and expected data sizes.
- A sentinel like `-999` may mean “missing”, not an actual negative measurement.
- Debugging order: identify impossible output → inspect min/max → inspect individual values → find source → clean or flag bad data → rerun.
- Removing invalid observations changes the dataset; report how many you dropped.


````python
scores = {"Amy":8.5,"Ben":7.0}
def check_scores(scores):
    for name in scores:
        score = scores[name]
        assert isinstance(score,(int,float)), f"{name}: score is not numeric"
        assert 0 <= score <= 10, f"{name}: score outside 0 to 10"
    return True
print(check_scores(scores))

dirty = {"Amy":8.5,"Ben":-999,"Cal":7.0}
clean = {}
for name in dirty:
    if dirty[name] >= 0:
        clean[name] = dirty[name]
print(clean)
print("Dropped:",len(dirty)-len(clean))
````
